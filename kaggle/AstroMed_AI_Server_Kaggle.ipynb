```python
# 3. V5.4 adapterini topish (Add Input orqali qo‘shilgan notebook chiqishidan)
import glob, shutil, zipfile
models = pathlib.Path('/kaggle/working/models'); models.mkdir(exist_ok=True)
cfgs = glob.glob('/kaggle/input/**/adapter_config.json', recursive=True)
if cfgs:
    src = pathlib.Path(cfgs[0]).parent
    shutil.copytree(src, models / 'adapter', dirs_exist_ok=True)
    for c in glob.glob('/kaggle/input/**/inference_config.json', recursive=True)[:1]:
        shutil.copy(c, models / 'inference_config.json')
else:
    zips = glob.glob('/kaggle/input/**/AstroMed_V5_RESULTS.zip', recursive=True)
    assert zips, 'Adapter topilmadi: Add Input → notebook167853da03 (V5.4 chiqishi) ni qo‘shing'
    with zipfile.ZipFile(zips[0]) as z:
        for name in z.namelist():
            if name.endswith('/'): continue
            parts = pathlib.Path(name).parts
            if 'adapter' in parts:
                target = models / 'adapter' / pathlib.Path(*parts[parts.index('adapter') + 1:])
            elif parts[-1] == 'inference_config.json':
                target = models / 'inference_config.json'
            else: continue
            target.parent.mkdir(parents=True, exist_ok=True)
            target.write_bytes(z.read(name))
print(sorted(p.name for p in (models / 'adapter').iterdir()))
assert (models / 'adapter' / 'adapter_config.json').is_file()
```

```python
# 4. Asosiy modellar (Kaggle internetida tez yuklanadi)
from huggingface_hub import snapshot_download
keep = ['*.json', '*.safetensors', '*.model', '*.txt', 'tokenizer*']
snapshot_download('Qwen/Qwen2.5-3B-Instruct', local_dir=str(models / 'qwen2.5-3b-instruct'), allow_patterns=keep)
snapshot_download('facebook/nllb-200-distilled-600M', local_dir=str(models / 'nllb-200-distilled-600M'), allow_patterns=keep + ['*.bin'])
print('Modellar tayyor')
```

```python
# 5. Modelni GPU'ga yuklash va sinov
import time
from astro_medical.ai_engine import AIEngine, DEMO_RECORDS
engine = AIEngine()
print(engine.status())
t = time.time(); engine.load(); print(f'Yuklandi: {time.time() - t:.0f} s')
r = engine.analyze(DEMO_RECORDS)
print('Harakat:', r['harakat'], '| vaqt:', r['sekund'], 's')
for k, v in r['qaydlar'].items(): print(k, v['sinf'], v['ishonch'])
```

```python
# 6. Serverni ishga tushirish va internetga ochish — shu katak ishlab tursin
import re, subprocess, threading, uvicorn
from astro_medical.ai_remote import create_ai_server
app = create_ai_server(engine)
threading.Thread(target=lambda: uvicorn.run(app, host='127.0.0.1', port=8000, log_level='warning'), daemon=True).start()
time.sleep(3)
tunnel = subprocess.Popen(['/kaggle/working/cloudflared', 'tunnel', '--no-autoupdate', '--url', 'http://127.0.0.1:8000'],
                          stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
url = None
for line in tunnel.stdout:
    m = re.search(r'https://[a-z0-9-]+\.trycloudflare\.com', line)
    if m: url = m.group(0); break
threading.Thread(target=lambda: [None for _ in tunnel.stdout], daemon=True).start()
print('=' * 70)
print('AI SERVER MANZILI:', url)
print('Shu manzilni AstroVital AstroAI panelidagi “Kaggle GPU ulanishi” maydoniga qo‘ying.')
print('=' * 70)
while True:  # session ochiq turadi; to‘xtatish uchun Stop bosing
    time.sleep(300)
    print(time.strftime('%H:%M'), 'server ishlayapti', url, flush=True)
```
